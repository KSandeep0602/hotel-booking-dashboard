
# Hotel Booking Dataset – Data Cleaning

## 1. Introduction

- Detecting and evaluating duplicate records.
- Detecting and evaluating numerical outliers.

Each cleaning decision will be documented and justified.


## 2. Duplicate Identification and Handling

Duplicate records were identified using Pandas' duplicated() method.

The initial assessment found 31,994 exact duplicate rows across all columns.

Since these rows have identical values across every available column, they may represent repeated records. As the dataset does not contain a unique booking identifier, exact duplicate removal is performed as a data-quality assumption.

The first occurrence of each identical record is retained.


```python

# Count duplicate rows before cleaning
duplicate_count = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicate_count)

# Display some duplicate records
duplicate_samples = df[df.duplicated(keep=False)]

display(duplicate_samples.head(10))

# Remove exact duplicate rows
df = df.drop_duplicates(keep="first").copy()

# Check dataset after duplicate removal
print("Dataset shape after removing duplicates:", df.shape)

print("Remaining duplicate rows:", df.duplicated().sum())
```

```python

# Complete missing-value report
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

missing_report = missing_report[
    missing_report["Missing Count"] > 0
].sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_report)

```


### Duplicate Handling – Decision

The original dataset contained 31,994 exact duplicate rows.
I inspected sample records and found rows with identical values
across all available columns.

This reduced the dataset from 119,390 to 87,396 rows.

### Missing Value Handling – Decisions

The dataset contained missing values in four columns:

- **children:** Only 4 values were missing. These were replaced with 0, assuming that the missing child count represents no children.
- **country:** 452 values were missing. These were replaced with "Unknown" to preserve the booking records without guessing the country.
- **agent:** 12,193 values were missing. These were replaced with "Unknown" because the agent column represents an identifier, and using the mean or median would not be meaningful.
- **company:** 82,137 values were missing (approximately 93.98%). This column was removed because such a large proportion of its values were missing, making it less useful for this analysis.

No rows were removed solely because they contained missing values. These decisions help preserve useful booking information while handling missing data appropriately.

```python
# Save a copy before handling missing values
df_cleaning_before = df.copy()

# 1. Fill missing children values with the median
children_median = df["children"].median()
df["children"] = df["children"].fillna(children_median)

# 2. Fill missing country values
df["country"] = df["country"].fillna("Unknown")

# 3. Fill missing agent values
df["agent"] = df["agent"].fillna("Unknown")

# 4. Drop company because approximately 94% of its values are missing
df = df.drop(columns=["company"])

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDataset shape after missing-value handling:")
print(df.shape)
```

```python
# Convert agent to a string column before filling missing values
df["agent"] = df["agent"].astype("string").fillna("Unknown")

print("Agent column data type:", df["agent"].dtype)
print("Missing agent values:", df["agent"].isnull().sum())
```

```python
print("Final dataset shape:", df.shape)

print("\nColumns with missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nChildren median used:", children_median)

print("\nRemaining duplicate rows:", df.duplicated().sum())
```

```python
# Find all rows involved in duplicate groups
duplicate_mask = df.duplicated(keep=False)

print("Number of rows involved in duplicate groups:",
      duplicate_mask.sum())

# Display duplicates after cleaning
display(df[duplicate_mask].sort_index())

# Compare these records with the data before missing-value handling
duplicate_indices = df.index[duplicate_mask]

print("\nOriginal records before missing-value handling:")
display(df_cleaning_before.loc[duplicate_indices])
```

## Outlier Detection and Handling

Outliers are unusually high or low values in numerical columns.

The Interquartile Range (IQR) method is used to identify potential outliers.

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are considered potential outliers. They will be inspected before deciding whether they should be removed or retained.

```python
# Save a copy before outlier handling
df_before_outliers = df.copy()

# Numerical columns selected for outlier analysis
numerical_columns = [
    "lead_time",
    "adults",
    "children",
    "babies",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "previous_cancellations",
    "previous_bookings_not_canceled",
    "booking_changes",
    "days_in_waiting_list",
    "adr",
    "required_car_parking_spaces",
    "total_of_special_requests"
]

# Calculate IQR boundaries and outlier counts
outlier_results = []

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_results.append({
        "Column": column,
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": outlier_count
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df)
```

```python
# 1. Inspect ADR values
print("ADR below zero:")
display(df[df["adr"] < 0][["hotel", "adults", "children", "adr"]])

print("\nADR above 500:")
display(
    df[df["adr"] > 500][
        ["hotel", "adults", "children", "adr",
         "stays_in_weekend_nights", "stays_in_week_nights"]
    ].sort_values("adr", ascending=False).head(15)
)

# 2. Inspect unusually high adult counts
print("\nBookings with more than 10 adults:")
display(
    df[df["adults"] > 10][
        ["hotel", "adults", "children", "babies", "adr"]
    ].sort_values("adults", ascending=False).head(15)
)

# 3. Inspect long lead times
print("\nBookings with lead time above 500 days:")
display(
    df[df["lead_time"] > 500][
        ["hotel", "lead_time", "is_canceled", "adults", "adr"]
    ].sort_values("lead_time", ascending=False).head(15)
)

# 4. Check the overall ranges
print("\nMinimum and maximum values:")
display(df[numerical_columns].agg(["min", "max"]).T)
```

### Outlier Handling – Decisions

The IQR method was used to identify potential outliers in numerical columns.

After inspecting the records, the following decisions were made:

1. **Adults:** Records with more than 10 adults were retained because they belonged to group bookings. These records were not considered confirmed data errors.

2. **ADR:** The record with ADR = -6.38 was removed because a negative room rate is invalid for this analysis.

3. **Extreme ADR:** The record with ADR = 5400 was removed because it was an extreme value for a one-night canceled booking and appeared anomalous.

4. **Other high ADR values:** Values of 508 and 510 were retained because their booking details did not establish that they were incorrect.

5. **Lead time:** Long lead-time bookings were retained because customers can make reservations many months in advance.

IQR was used as an identification method, not as an automatic deletion rule. Only records with a clear justification were removed.

```python
# Check the number of records before outlier handling
print("Rows before outlier handling:", len(df))

# Count invalid and extreme ADR records
negative_adr_count = (df["adr"] < 0).sum()
extreme_adr_count = (df["adr"] > 1000).sum()

print("Negative ADR records:", negative_adr_count)
print("Extreme ADR records above 1000:", extreme_adr_count)

# Remove only the identified ADR anomalies
df = df[(df["adr"] >= 0) & (df["adr"] <= 1000)].copy()

# Check the result
print("\nRows after outlier handling:", len(df))
print("Remaining negative ADR values:", (df["adr"] < 0).sum())
print("Remaining ADR values above 1000:", (df["adr"] > 1000).sum())
```

### Outlier Handling – Results

After inspecting potential outliers using the IQR method, I removed two records:

- One record with a negative ADR value (-6.38), because a negative room rate is invalid for this analysis.
- One record with an extreme ADR value (5400), which appeared anomalous after inspecting its booking details.

Other high ADR values (508 and 510) were retained because there was not enough evidence to classify them as incorrect.

Long lead times and large group bookings were also retained because they can represent legitimate reservations.

### Categorical Encoding – Decisions

Categorical columns were handled according to their number of unique values and meaning.

- Low-cardinality columns such as hotel, meal, market_segment, distribution_channel, reserved_room_type, assigned_room_type, deposit_type, customer_type, and reservation_status were selected for one-hot encoding.
- arrival_date_month was converted into numerical month values from 1 to 12, preserving calendar order.
- country and agent were retained as categorical identifiers because they contain many unique values.
- reservation_status_date was converted into datetime format rather than one-hot encoded.

A separate encoded dataset was created so that the original cleaned dataset remains readable.

Note: reservation_status should be excluded from input features when predicting booking cancellations because it can reveal the booking outcome.

```python
# Create a separate copy for encoding
df_encoded = df.copy()

# Convert month names into numerical month values
month_mapping = {
    "January": 1, "February": 2, "March": 3,
    "April": 4, "May": 5, "June": 6,
    "July": 7, "August": 8, "September": 9,
    "October": 10, "November": 11, "December": 12
}

df_encoded["arrival_date_month"] = (
    df_encoded["arrival_date_month"].map(month_mapping)
)

# Convert reservation date into datetime format
df_encoded["reservation_status_date"] = pd.to_datetime(
    df_encoded["reservation_status_date"],
    errors="coerce"
)

# Select columns for one-hot encoding
columns_to_encode = [
    "hotel",
    "meal",
    "market_segment",
    "distribution_channel",
    "reserved_room_type",
    "assigned_room_type",
    "deposit_type",
    "customer_type",
    "reservation_status"
]

# Apply one-hot encoding
df_encoded = pd.get_dummies(
    df_encoded,
    columns=columns_to_encode,
    dtype=int
)

# Display results
print("Original cleaned dataset shape:", df.shape)
print("Encoded dataset shape:", df_encoded.shape)

display(df_encoded.head())
```

### Categorical Encoding – Results

One-hot encoding and appropriate data-type conversions were successfully applied.

The original cleaned dataset contained 87,394 rows and 31 columns. After encoding, the separate encoded dataset contained 87,394 rows and 74 columns.

The number of rows remained unchanged, while the number of columns increased because categorical values were converted into numerical columns.

The original cleaned dataset was preserved in `df`, and the encoded version was stored in `df_encoded`.
